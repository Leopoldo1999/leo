```python
# about the MC desition aproach
import numpy as np
import random
import matplotlib.pyplot as plt

pab=0.5
pcb=0.5
pcd=0.1
pad=0.9

#a=sunny tomorrow
#d=sunny today
#b=cloudy today
#c=cloudy tomorrow
d=1
sunny=[]
N = 3000
for i in range(N):
    if i == 0: #the first day
        if d == 1:
            sunny.append(random.choices([1, 0], weights=[pad, pcd])[0]) #if today sunny, choose tomorrow based on pad and pcd
        else:
            sunny.append(random.choices([1, 0], weights=[pab, pcb])[0]) #if today is cloudy choose based on pab and pcb
    else:
        if sunny[i - 1] == 1: #based on yesterday 
            sunny.append(random.choices([1, 0], weights=[pad, pcd])[0])
        else:
            sunny.append(random.choices([1, 0], weights=[pab, pcb])[0])
            
sunny_fraction=[]
cloudy_fraction=[]
for i in range(N):
    sunny_fraction.append((sum(sunny[0:i]))/(i+1))
    cloudy_fraction.append(1-(sum(sunny[0:i]))/(i+1))

plt.plot(np.arange(0,N), sunny_fraction, color='b', label="Sunny fraction")
plt.plot(np.arange(0,N), cloudy_fraction, color='r', label="Cloudy fraction")
plt.xlabel('Days')
plt.ylabel('Prob')
plt.legend();
```

```python
plt.plot(np.arange(200,N), sunny_fraction[200:], color='b', label="Sunny fraction")
plt.plot(np.arange(200,N), cloudy_fraction[200:], color='r', label="Cloudy fraction")
plt.xlabel('Days')
plt.ylabel('Prob')
plt.legend();
#IT WILL BE STABLE 10% BRUN IN?
```
